# Recommender systems and cosine similarity

Similarity is measured using the distance metric. Nearest points are the most similar and farthest points are the least relevant. The similarity is subjective and is highly dependent on the domain and application. For example, two movies are similar because of genre or length or cast. Care should be taken when calculating distance across dimensions/features that are unrelated. The relative values of each element must be normalized, or one feature could end up dominating the distance calculation.

Cosine similarity measures the cosine of the angle between two vectors. It is a judgment of orientation rather than magnitude between two vectors with respect to the origin. The cosine of 0 degrees is 1, which means the data points are similar. The cosine of 90 degrees is 0, which means data points are not similar. Cosine similarity is not an actual distance metric, since it is subjective to the domain and application. For example data points \[1,2\] and \[10,20\], are shown as similar with cosine similarity, while the Euclidean distance measure shows them as being far away from each other.



```python
import numpy as np
import pandas as pd
import nltk

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv('tmdb_movies.csv')
```

```python
df.head(5)
```

```python
# set max_features to the size of the dataset

FEATURES = 1000  # 750 to 1000 features

count_vectorizer = CountVectorizer(max_features=FEATURES, stop_words='english')
```

```python
vectors = count_vectorizer.fit_transform(df['attributes']).toarray()

print ("vectors shape", vectors.shape)
```

```python
# similarity is a matriz of vectors

similarity = cosine_similarity(vectors)

print ("similarity shape", similarity.shape)
```

```python
# returns a list containing up to 5 recommended movies

def recommend(title):
    try:
        index = df[df['title'] == title].index[0]

        distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])

        output = []

        for item in distances[1:6]:
            output.append(df.iloc[item[0]].title)

        return output
    except:
        print(title,"is not in the dataset")
```

```python
# use the recomend function with a movie title in the dataset to see what other movies are recomended. you can change the number recomended in the "recomend" function above.
```

```python
recommend('Transformers')
```

```python
recommend('TRON: Legacy')
```

```python
recommend("Monsters, Inc.")
```
